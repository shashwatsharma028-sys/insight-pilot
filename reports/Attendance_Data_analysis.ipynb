## Descriptive Statistics of Attendance Metrics
*Calculate mean, median, and distribution of hours_worked and base_salary to establish a baseline for employee productivity and compensation.*

```
CHART_PATH = 'chart_task_001.png'

import matplotlib.pyplot as plt
import pandas as pd

# Clean data for analysis
df_clean = df.dropna(subset=['hours_worked', 'base_salary'])

# Calculate descriptive statistics
stats = df_clean[['hours_worked', 'base_salary']].agg(['mean', 'median', 'std', 'min', 'max'])

print("Descriptive Statistics for Attendance Metrics:")
print(stats)

# Plotting distributions
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(df_clean['hours_worked'], bins=20, color='skyblue', edgecolor='black')
axes[0].set_title('Distribution of Hours Worked')
axes[0].set_xlabel('Hours')
axes[0].set_ylabel('Frequency')

axes[1].hist(df_clean['base_salary'], bins=20, color='salmon', edgecolor='black')
axes[1].set_title('Distribution of Base Salary')
axes[1].set_xlabel('Salary')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.tight_layout()
plt.show()

# Key findings
print(f"Mean hours worked: {df_clean['hours_worked'].mean():.2f} hours")
print(f"Median hours worked: {df_clean['hours_worked'].median():.2f} hours")
print(f"Mean base salary: ${df_clean['base_salary'].mean():,.2f}")
print(f"Median base salary: ${df_clean['base_salary'].median():,.2f}")
```

**Insight:** The analysis of 15,600 attendance records shows a mean of 8.07 hours worked per shift, with a median of 8.70 hours. The base salary data indicates a mean of $20,377.16 and a median of $20,000.00 across the workforce.

*Confidence: 95%*

## Outlier Detection in Hours Worked
*Identify and analyze the 1271 outliers in hours_worked to determine if these represent data entry errors or extreme work shifts.*

```
CHART_PATH = 'chart_task_002.png'

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df['hours_worked'] = pd.to_numeric(df['hours_worked'], errors='coerce')
clean_df = df.dropna(subset=['hours_worked'])

q1 = clean_df['hours_worked'].quantile(0.25)
q3 = clean_df['hours_worked'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = clean_df[(clean_df['hours_worked'] < lower_bound) | (clean_df['hours_worked'] > upper_bound)]

print(f"Total records analyzed: {len(clean_df)}")
print(f"Outlier threshold: < {lower_bound:.2f} or > {upper_bound:.2f} hours")
print(f"Number of outliers detected: {len(outliers)}")
print(f"Mean hours of outliers: {outliers['hours_worked'].mean():.2f}")
print(f"Percentage of total data that are outliers: {(len(outliers)/len(clean_df))*100:.2f}%")

dept_outliers = outliers.groupby('department').size().sort_values(ascending=False)
print("\nTop departments with outliers:")
print(dept_outliers.head())

plt.figure(figsize=(10, 6))
plt.hist(clean_df['hours_worked'], bins=30, color='skyblue', alpha=0.7, label='Normal Data')
plt.hist(outliers['hours_worked'], bins=30, color='red', alpha=0.7, label='Outliers')
plt.title('Distribution of Hours Worked: Outliers vs Normal')
plt.xlabel('Hours Worked')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```

**Insight:** The analysis identified 1,271 outliers in hours_worked, representing 8.31% of the 15,288 records analyzed. These outliers fall outside the 5.10 to 12.30-hour range, with a mean value of only 0.13 hours. The Maintenance department accounts for the highest volume of these anomalies with 431 instances, followed by Reception with 332.

*Confidence: 95%*

## Departmental Productivity Comparison
*Compare average hours_worked across different departments to identify potential staffing or operational inefficiencies.*

```
CHART_PATH = 'chart_task_003.png'

import pandas as pd
import matplotlib.pyplot as plt

try:
    df_clean = df.dropna(subset=['department', 'hours_worked'])
    dept_productivity = df_clean.groupby('department')['hours_worked'].mean().sort_values(ascending=False)
    
    print("Average hours worked per department:")
    for dept, avg in dept_productivity.items():
        print(f"{dept}: {avg:.2f} hours")
        
    top_dept = dept_productivity.idxmax()
    bottom_dept = dept_productivity.idxmin()
    print(f"\nMost productive department (by avg hours): {top_dept} ({dept_productivity.max():.2f} hrs)")
    print(f"Least productive department (by avg hours): {bottom_dept} ({dept_productivity.min():.2f} hrs)")

    plt.figure(figsize=(10, 6))
    dept_productivity.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Average Hours Worked by Department')
    plt.xlabel('Department')
    plt.ylabel('Average Hours Worked')
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    
    plt.savefig(CHART_PATH, bbox_inches='tight')
    plt.close('all')

except Exception as e:
    print(f"Error during analysis: {e}")
```

**Insight:** The analysis of 15,600 attendance records shows a high degree of consistency in average hours worked across departments, ranging from 8.04 hours in Maintenance to 8.10 hours in Housekeeping. The difference between the highest and lowest performing departments is only 0.06 hours per shift.

*Confidence: 95%*

## Daily Attendance Trend Analysis
*Analyze the trend of hours_worked over the date column to identify seasonal patterns or fluctuations in workforce attendance.*

```
CHART_PATH = 'chart_task_004.png'

import pandas as pd
import matplotlib.pyplot as plt

try:
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df_clean = df.dropna(subset=['date', 'hours_worked'])
    
    daily_trend = df_clean.groupby('date')['hours_worked'].mean().sort_index()
    
    plt.figure(figsize=(12, 6))
    plt.plot(daily_trend.index, daily_trend.values, marker='o', linestyle='-', markersize=4)
    plt.title('Daily Average Hours Worked Trend')
    plt.xlabel('Date')
    plt.ylabel('Average Hours Worked')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.savefig(CHART_PATH, bbox_inches='tight')
    plt.close('all')
    
    overall_mean = daily_trend.mean()
    max_day = daily_trend.idxmax()
    max_val = daily_trend.max()
    min_day = daily_trend.idxmin()
    min_val = daily_trend.min()
    
    print(f"Overall average daily hours worked: {overall_mean:.2f}")
    print(f"Highest average attendance day: {max_day.strftime('%Y-%m-%d')} ({max_val:.2f} hours)")
    print(f"Lowest average attendance day: {min_day.strftime('%Y-%m-%d')} ({min_val:.2f} hours)")
    
except Exception as e:
    print(f"Error during trend analysis: {e}")
```

**Insight:** The average daily hours worked across the dataset is 8.07 hours. Daily attendance peaked on 2024-02-15 with an average of 8.52 hours, while the lowest attendance occurred on 2024-01-18 with an average of 7.50 hours.

*Confidence: 95%*

## Correlation Between Salary and Hours
*Evaluate the correlation between base_salary and hours_worked to determine if compensation levels align with actual time spent on site.*

```
CHART_PATH = 'chart_task_005.png'

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

try:
    df_clean = df.dropna(subset=['hours_worked', 'base_salary'])
    
    correlation = df_clean['hours_worked'].corr(df_clean['base_salary'])
    
    print(f"Correlation between base_salary and hours_worked: {correlation:.4f}")
    print(f"Number of observations used: {len(df_clean)}")
    
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df_clean, x='hours_worked', y='base_salary', alpha=0.5)
    plt.title('Correlation: Base Salary vs Hours Worked')
    plt.xlabel('Hours Worked')
    plt.ylabel('Base Salary')
    plt.grid(True, linestyle='--', alpha=0.7)
    
    plt.savefig(CHART_PATH, bbox_inches='tight')
    plt.close('all')
    
except Exception as e:
    print(f"An error occurred during analysis: {e}")
```

**Insight:** The correlation coefficient between base_salary and hours_worked is 0.0004 across 15,288 observations. This indicates there is effectively zero linear relationship between an employee's base salary and the number of hours they work.

*Confidence: 95%*
