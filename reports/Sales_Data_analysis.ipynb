## Descriptive Statistics Overview
*Calculate mean, median, and variance for sales_amount, units_sold, and customer_satisfaction to establish a performance baseline.*

```
CHART_PATH = 'chart_task_001.png'

import pandas as pd
import matplotlib.pyplot as plt

cols_to_analyze = ['sales_amount', 'units_sold', 'customer_satisfaction']

stats = df[cols_to_analyze].agg(['mean', 'median', 'var'])

print("Descriptive Statistics Overview:")
for col in cols_to_analyze:
    print(f"\n--- {col} ---")
    print(f"Mean:   {stats.loc['mean', col]:,.2f}")
    print(f"Median: {stats.loc['median', col]:,.2f}")
    print(f"Var:    {stats.loc['var', col]:,.2f}")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for i, col in enumerate(cols_to_analyze):
    df[col].hist(ax=axes[i], bins=15, color='skyblue', edgecolor='black')
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

plt.tight_layout()
plt.tight_layout()
plt.show()
```

## Sales Performance Trend Analysis
*Analyze the monthly trend of sales_amount and units_sold over time to identify seasonal growth patterns.*

```
CHART_PATH = 'chart_task_002.png'

import pandas as pd
import matplotlib.pyplot as plt

try:
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df_clean = df.dropna(subset=['date', 'sales_amount', 'units_sold'])
    
    monthly_trend = df_clean.groupby(df_clean['date'].dt.to_period('M')).agg({
        'sales_amount': 'sum',
        'units_sold': 'sum'
    }).sort_index()

    print("Monthly Sales Performance Summary:")
    print(monthly_trend)
    
    total_sales = monthly_trend['sales_amount'].sum()
    peak_month = monthly_trend['sales_amount'].idxmax()
    print(f"Total Sales: ${total_sales:,.0f}")
    print(f"Peak Sales Month: {peak_month}")

    fig, ax1 = plt.subplots(figsize=(10, 6))

    color = 'tab:blue'
    ax1.set_xlabel('Date')
    ax1.set_ylabel('Sales Amount ($)', color=color)
    ax1.plot(monthly_trend.index.astype(str), monthly_trend['sales_amount'], color=color, marker='o', label='Sales Amount')
    ax1.tick_params(axis='y', labelcolor=color)
    ax1.tick_params(axis='x', rotation=45)

    ax2 = ax1.twinx()
    color = 'tab:red'
    ax2.set_ylabel('Units Sold', color=color)
    ax2.plot(monthly_trend.index.astype(str), monthly_trend['units_sold'], color=color, marker='s', linestyle='--', label='Units Sold')
    ax2.tick_params(axis='y', labelcolor=color)

    plt.title('Monthly Sales Amount and Units Sold Trend')
    plt.savefig(CHART_PATH, bbox_inches='tight')
    plt.close('all')

except Exception as e:
    print(f"Error during trend analysis: {e}")
```

**Insight:** Monthly sales peaked in May 2023 at $1,561,496, which is more than double the average monthly sales of approximately $848,370. There is a notable volatility in performance, with sales dropping from $1,561,496 in May 2023 to $706,310 in June 2023, and a significant single-month decline to $407,045 in March 2024.

*Confidence: 95%*

## Correlation Analysis of Sales Drivers
*Examine the relationship between sales_amount, units_sold, and customer_satisfaction to determine if higher satisfaction correlates with higher revenue.*

```
CHART_PATH = 'chart_task_003.png'

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = df[['sales_amount', 'units_sold', 'customer_satisfaction']].dropna()

correlation_matrix = df_clean.corr()

sales_sat_corr = correlation_matrix.loc['sales_amount', 'customer_satisfaction']
units_sat_corr = correlation_matrix.loc['units_sold', 'customer_satisfaction']
sales_units_corr = correlation_matrix.loc['sales_amount', 'units_sold']

print(f"Correlation between Sales Amount and Customer Satisfaction: {sales_sat_corr:.3f}")
print(f"Correlation between Units Sold and Customer Satisfaction: {units_sat_corr:.3f}")
print(f"Correlation between Sales Amount and Units Sold: {sales_units_corr:.3f}")
print(f"Sample size (n) for correlation analysis: {len(df_clean)}")

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title('Correlation Heatmap: Sales Drivers')
plt.tight_layout()
plt.show()
```

**Insight:** The analysis shows a strong positive correlation of 0.843 between units sold and total sales amount, confirming that volume is the primary driver of revenue. Conversely, there is a negligible correlation between customer satisfaction and both sales amount (-0.002) and units sold (-0.057), indicating that current satisfaction scores do not track with sales performance.

*Confidence: 95%*

## Regional Performance Comparison
*Compare average sales_amount and return rates across different regions to identify high-performing vs. problematic market areas.*

```
CHART_PATH = 'chart_task_004.png'

import pandas as pd
import matplotlib.pyplot as plt

# Ensure numeric columns are clean
df['sales_amount'] = pd.to_numeric(df['sales_amount'], errors='coerce')
df['returns'] = pd.to_numeric(df['returns'], errors='coerce')

# Group by region
regional_stats = df.groupby('region').agg({
    'sales_amount': 'mean',
    'returns': 'mean'
}).reset_index()

# Print findings
print("Regional Performance Summary (Averages):")
for _, row in regional_stats.iterrows():
    print(f"Region: {row['region']} | Avg Sales: ${row['sales_amount']:,.2f} | Avg Returns: {row['returns']:.2f}")

# Identify top and bottom
top_region = regional_stats.loc[regional_stats['sales_amount'].idxmax()]
high_return_region = regional_stats.loc[regional_stats['returns'].idxmax()]

print(f"\nTop performing region by sales: {top_region['region']} (${top_region['sales_amount']:,.2f})")
print(f"Region with highest average returns: {high_return_region['region']} ({high_return_region['returns']:.2f} units)")

# Visualization
fig, ax1 = plt.subplots(figsize=(10, 6))

color = 'tab:blue'
ax1.set_xlabel('Region')
ax1.set_ylabel('Avg Sales Amount', color=color)
ax1.bar(regional_stats['region'], regional_stats['sales_amount'], color=color, alpha=0.6, label='Avg Sales')
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:red'
ax2.set_ylabel('Avg Returns', color=color)
ax2.plot(regional_stats['region'], regional_stats['returns'], color=color, marker='o', linewidth=2, label='Avg Returns')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Regional Performance: Average Sales vs Returns')
plt.tight_layout()
plt.show()
```

**Insight:** The North region leads in average sales at $52,750.92 per transaction, outperforming the South region, which has the lowest average sales at $40,081.79. However, the North region also reports the highest average returns at 7.23 units per transaction, compared to the South region's lowest return rate of 4.36 units.

*Confidence: 95%*

## Outlier Detection in Returns and Sales
*Investigate the 10 outliers in units_sold and 15 outliers in returns to determine if these represent data entry errors or specific product/rep performance anomalies.*

```
CHART_PATH = 'chart_task_005.png'

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def detect_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    return (series < (q1 - 1.5 * iqr)) | (series > (q3 + 1.5 * iqr))

df['units_outlier'] = detect_outliers(df['units_sold'])
df['returns_outlier'] = detect_outliers(df['returns'])

units_outliers = df[df['units_outlier']]
returns_outliers = df[df['returns_outlier']]

print(f"Total units_sold outliers: {len(units_outliers)}")
print(f"Total returns outliers: {len(returns_outliers)}")

print("\nTop products in units_sold outliers:")
print(units_outliers['product'].value_counts().head(3))

print("\nTop sales_reps in returns outliers:")
print(returns_outliers['sales_rep'].value_counts().head(3))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].scatter(df['units_sold'], df['sales_amount'], c=df['units_outlier'], cmap='coolwarm', alpha=0.6)
ax[0].set_title('Units Sold vs Sales Amount (Outliers Highlighted)')
ax[0].set_xlabel('Units Sold')
ax[0].set_ylabel('Sales Amount')

ax[1].scatter(df['returns'], df['sales_amount'], c=df['returns_outlier'], cmap='viridis', alpha=0.6)
ax[1].set_title('Returns vs Sales Amount (Outliers Highlighted)')
ax[1].set_xlabel('Returns')
ax[1].set_ylabel('Sales Amount')

plt.tight_layout()
plt.tight_layout()
plt.show()

correlation = df[['units_sold', 'returns', 'sales_amount']].corr()
print("\nCorrelation matrix for key metrics:")
print(correlation)
```

**Insight:** The analysis identified 10 outliers in units_sold, all of which are associated with Product C, and 15 outliers in returns, with Rep_14 accounting for 3 of these instances. There is a strong positive correlation of 0.77 between units_sold and returns, suggesting that higher volume sales are consistently linked to higher return rates.

*Confidence: 85%*
